EDA

How are ratings distributed?

```python
city_rating=df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False)
print(city_rating.head(10))
```

Which cities have the highest number of restaurants?

```python
city_count=df.groupby('City')['Restaurant Name'].count().sort_values(ascending=False)
print(city_count.head(10))
```

 What is the impact of online delivery on ratings?

```python
delivery_rating=df.groupby('Has Online delivery')['Aggregate rating'].mean()
print(delivery_rating)
```

```python
df['Has Online delivery'].value_counts()
```

```python
df['Average Cost for two'].value_counts()
```

```python
cost_rating=df.groupby('Average Cost for two')['Aggregate rating'].mean()
print(cost_rating)
```

VISUALIZATION


```python
city_rating.head(10).plot(kind='bar')
plt.title('Average ratings by city')
plt.show()
```

```python
delivery_rating.plot(kind='bar')
plt.title('Ratings by online delivery')
plt.show()
```

```python
sns.scatterplot(x='Average Cost for two',y='Aggregate rating',data=df)
plt.title('Cost vs Ratings')
plt.show()
```
